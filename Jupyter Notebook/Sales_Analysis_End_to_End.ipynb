```python
#importing libraries for accessing specific features
import pandas as pd  # for loading and handling csv data files
import numpy as np  # for numerical calculations like mean, sum etc
import matplotlib.pyplot as plt  # for bar charts and line charts
import seaborn as sns  # for designing charts

import warnings
warnings.filterwarnings('ignore')  # to hide unnecessary warning messages
```

```python
df = pd.read_csv('Raw Data.csv')

df.columns = df.columns.str.strip()

df['Discount Band'] = df['Discount Band'].fillna('None')
df.to_csv('cleaned_dataset.csv', index=False)

print("Done! Cleaned shape:", df.shape)
```

```python
# 3. Checking data - how big is data, any missing?
df.head()
df.info()
df.isnull().sum()  # checking the null values
print("duplicates:", df.duplicated().sum())
```

```python
# Profit by Segment - which segment gives more profit?
df.groupby('Segment')['Profit'].sum().sort_values(ascending=False)

# Sales by Country - which country sells most?
df.groupby('Country')['Sales'].sum().sort_values(ascending=False)

# Profit by Product
df.groupby('Product')['Profit'].sum().sort_values(ascending=False)

# Yearly - 2013 vs 2014 growth
df.groupby('Year')['Profit'].sum()
```

```python
# Charts for PowerBI Dashboard presentation
df.groupby('Segment')['Profit'].sum().plot(kind='barh')
plt.title('Profit by Segment')
plt.show()

df.groupby('Country')['Sales'].sum().sort_values().plot(kind='barh')
plt.title('Sales by Country')
plt.show()
```

```python
# final insights - for power bi report
print("1. Government segment best - 11M profit")
print("2. USA top country - 25M sales")
print("3. Profit grew 2013 to 2014 a lot")
print("4. Small Business also good")
```
